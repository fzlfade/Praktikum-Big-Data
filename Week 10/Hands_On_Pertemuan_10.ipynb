### 1. Menghubungkan ke Database MongoDB
- **Tugas 1**: Pastikan MongoDB telah terpasang dan berjalan. Hubungkan ke MongoDB lokal.


```python
from pymongo import MongoClient

# Inisialisasi client dan koneksi ke database
client = MongoClient('mongodb://localhost:27017/')
db = client['company_db']
collection = db['employees']

# Contoh untuk memastikan koneksi
print('Connected to MongoDB')
```

### 2. Operasi CRUD Dasar
- **Tugas 2**: Insert, Update, dan Delete data pada koleksi `employees`.


```python
# Contoh Insert Data
employee_data = {
    'name': 'Faizal',
    'department': 'Human Resources',
    'age': 21,
    'salary': 45000
}
collection.insert_one(employee_data)
print('Data inserted')

# Tugas: Insert beberapa data tambahan, lakukan update, serta delete data
```

```python
# Contoh Update Data
employee_data = {'salary': 45000}
new_values = {'$set': {'salary': 50000}}

collection.update_one(employee_data, new_values)
print('Data updated')
```

```python
# Contoh Delete Data
employee_data = {'name': 'Alice Wonderland'}

collection.delete_one(employee_data)
print('Data deleted')
```

```python
# Contoh Query Data
for employee in collection.find():
    print(employee)
```

### 3. Query Lebih Kompleks Menggunakan Aggregation
- **Tugas 3**: Terapkan aggregation untuk menghitung rata-rata gaji per departemen.


```python
# Query Aggregation untuk mencari rata-rata gaji per departemen
pipeline = [
    {'$group': {'_id': '$department', 'average_salary': {'$avg': '$salary'}}}
]
for result in collection.aggregate(pipeline):
    print(result)
```

### 4. Latihan Tambahan
- **Latihan 4.1**: Buatlah koleksi baru `products` dan masukkan data produk (minimal 10 produk).
- **Latihan 4.2**: Lakukan query untuk menemukan produk dengan harga di atas rata-rata.
- **Latihan 4.3**: Buatlah aggregation pipeline untuk menghitung total produk dalam setiap kategori.


```python
products_collection = db['products']
products_data = [
    {'name': 'Apple', 'price': 1, 'category': 'Fruits', 'quantity': 200},
    {'name': 'Beef', 'price': 20, 'category': 'Meat', 'quantity': 50},
    {'name': 'Chicken Nugget', 'price': 10, 'category': 'Meat', 'quantity': 1000},
    {'name': 'Donut', 'price': 1, 'category': 'Bakery', 'quantity': 100},
    {'name': 'Egg', 'price': 0.5, 'category': 'Dairy', 'quantity': 300},
    {'name': 'Fish', 'price': 3, 'category': 'Seafood', 'quantity': 50},
    {'name': 'Grapes', 'price': 0.1, 'category': 'Fruits', 'quantity': 1000},
    {'name': 'Honey', 'price': 2, 'category': 'Condiments', 'quantity': 100},
    {'name': 'Instant Noodles', 'price': 1, 'category': 'Pantry', 'quantity': 2000},
    {'name': 'Jelly', 'price': 2, 'category': 'Desserts', 'quantity': 500}
]

products_collection.insert_many(products_data)
print('Data inserted')
```

```python
for products in collection.find():
    print(products)
```

```python
average_price = products_collection.aggregate([
    {"$group": {"_id": None, "averagePrice": {"$avg": "$price"}}}
]).next()["averagePrice"]

expensive_products = products_collection.find({"price": {"$gt": average_price}})
for product in expensive_products:
    print(product)
```

```python
pipeline = [
    {
        "$group": {
            "_id": "$category",
            "totalQuantity": {"$sum": "$quantity"}
        }
    },
    {
        "$project": {
            "category": "$_id",
            "totalQuantity": 1,
            "_id": 0
        }
    }
]

for result in products_collection.aggregate(pipeline):
    print(result)
```

### 5. Tugas
- **Tugas 1**: Cari 5 karyawan dengan gaji tertinggi dalam setiap departemen, gunakan query atau aggregation yang sesuai.
- **Tugas 2**: Buatlah skenario di mana Anda harus menghapus karyawan yang berusia di bawah 25 tahun dari database.
- **Tugas 3**: Buatlah laporan ringkas (menggunakan MongoDB query) yang menghitung total gaji karyawan di setiap departemen, serta rata-rata umur karyawan.


### Tugas 1

```python
pipeline = [
    {
        "$sort": {"department": 1, "salary": -1}
    },
    {
        "$group": {
            "_id": "$department",
            "topEmployees": {
                "$push": {
                    "name": "$name",
                    "salary": "$salary"
                }
            }
        }
    },
    {
        "$project": {
            "topEmployees": {"$slice": ["$topEmployees", 5]}
        }
    }
]

for result in collection.aggregate(pipeline):
    print(result)
```
