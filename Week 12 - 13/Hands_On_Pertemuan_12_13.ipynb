## Pertemuan 12: Data Cleaning and Preparation using Pandas

### Topics Covered

- Data transformation and normalization.
- Data filtering and deduplication.
- Standardization of categorical data.
- Outlier detection and handling.


```python
# Exercise 1: Identifying and Handling Missing Data
import pandas as pd

print(pd.__version__)

# Sample dataset with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', None],
    'Age': [24, 30, None, 22, 35],
    'Salary': [48000, None, 57000, None, 60000]
}
df = pd.DataFrame(data)

# Filling missing values and dropping rows
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].median(), inplace=True)
df.dropna(subset=['Name'], inplace=True)
print('After cleaning:\n', df)
```

```python
# Exercise 2: Standardizing Categorical Data
# Sample dataset with inconsistent categorical values
data = {
    'Product': ['Laptop', 'Laptop', 'Desktop', 'Tablet', 'Tablet'],
    'Category': ['Electronics', 'electronics', 'Electronics', 'Gadgets', 'gadgets']
}
df = pd.DataFrame(data)

# Standardize category values
df['Category'] = df['Category'].str.capitalize()
print('Standardized Data:\n', df)
```

#  Use Matplotlib to create a multi-bar chart to compare data across groups.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load the dataset
df = pd.read_csv('banana_quality_dataset.csv')

# Calculate average quality score by variety and region
avg_quality = df.groupby(['variety', 'region'])['quality_score'].mean().unstack()

# Plotting
fig, ax = plt.subplots(figsize=(12, 6))
x = np.arange(len(avg_quality.index))  # the label locations
width = 0.15  # the width of the bars

# Create bars for each region
for i, region in enumerate(avg_quality.columns):
    ax.bar(x + i*width, avg_quality[region], width, label=region)

# Add some text for labels, title and custom x-axis tick labels, etc.
ax.set_xlabel('Variety')
ax.set_ylabel('Average Quality Score')
ax.set_title('Average Quality Score by Variety and Region')
ax.set_xticks(x + width * (len(avg_quality.columns) - 1) / 2)
ax.set_xticklabels(avg_quality.index, rotation=45)
ax.legend()

fig.tight_layout()
plt.show()
```

```python
import seaborn as sns

# Scatter plot to explore relationship between ripeness_index and sugar_content_brix
sns.scatterplot(data=df, x='ripeness_index', y='sugar_content_brix', hue='variety')
plt.title('Scatter Plot of Ripeness Index vs Sugar Content Brix')
plt.show()
```

```python
# Customizing Matplotlib plot
fig, ax = plt.subplots(figsize=(12, 6))
for i, region in enumerate(avg_quality.columns):
    ax.bar(x + i*width, avg_quality[region], width, label=region, color=plt.cm.tab20(i))

ax.set_xlabel('Variety')
ax.set_ylabel('Average Quality Score')
ax.set_title('Average Quality Score by Variety and Region')
ax.set_xticks(x + width * (len(avg_quality.columns) - 1) / 2)
ax.set_xticklabels(avg_quality.index, rotation=45)
ax.legend()

fig.tight_layout()
plt.show()

# Customizing Seaborn plot
sns.set(style='whitegrid', palette='muted')
sns.scatterplot(data=df, x='ripeness_index', y='sugar_content_brix', hue='variety', palette='deep')
plt.title('Scatter Plot of Ripeness Index vs Sugar Content Brix')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler

# Load the dataset
df = pd.read_csv('music.csv')

# Check for missing values
print(df.isnull().sum())

# Fill missing values with 0 (assuming missing values mean no preference)
df.fillna(0, inplace=True)

# Normalize the data
scaler = MinMaxScaler()
df_normalized = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)

# Handle outliers by capping them at the 99th percentile
for col in df_normalized.columns:
    upper_limit = df_normalized[col].quantile(0.99)
    df_normalized[col] = np.where(df_normalized[col] > upper_limit, upper_limit, df_normalized[col])

# Display the cleaned and normalized data
print(df_normalized.describe())
```

```python
# Calculate the correlation matrix
corr_matrix = df_normalized.corr()

# Plotting the heatmap
plt.figure(figsize=(12, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Genre Preferences')
plt.show()

# Calculate the correlation matrix
corr_matrix = df_normalized.corr()

# Plotting the heatmap
plt.figure(figsize=(12, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Genre Preferences')
plt.show()

# Plotting the pairplot
sns.pairplot(df_normalized)
plt.suptitle('Pairplot of Normalized Genre Preferences', y=1.02)
plt.show()

```

```python
import matplotlib.gridspec as gridspec

# Calculate the sum of preferences for each genre
genre_sums = df_normalized.sum()

# Create a figure with a grid layout
fig = plt.figure(figsize=(18, 12))
gs = gridspec.GridSpec(2, 2, figure=fig)

# Bar Chart
ax1 = fig.add_subplot(gs[0, 0])
genre_sums.plot(kind='bar', color='skyblue', ax=ax1)
ax1.set_title('Total Preferences for Each Genre')
ax1.set_xlabel('Genre')
ax1.set_ylabel('Total Preferences')
ax1.set_xticklabels(ax1.get_xticklabels(), rotation=45)

# Heatmap
ax2 = fig.add_subplot(gs[0, 1])
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5, ax=ax2)
ax2.set_title('Correlation Matrix of Genre Preferences')

# Pairplot
sns.pairplot(df_normalized)
plt.suptitle('Pairplot of Normalized Genre Preferences', y=1.02)

plt.tight_layout()
plt.show()
```
